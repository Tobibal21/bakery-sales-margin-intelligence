```python
# ============================================================
# STAGE 5 - BUSINESS INSIGHTS & KPI ANALYSIS
# ============================================================

import pandas as pd

PROCESSED_PATH = r"C:\Users\user\Desktop\Bakery dashboard\data\processed"

# Load Stage 4 profitability data
Profitability = pd.read_csv(
    PROCESSED_PATH + r"\stage4_profitability.csv"
)

print("=" * 60)
print("STAGE 5 - BUSINESS INSIGHTS & KPI ANALYSIS")
print("=" * 60)

# ============================================================
# 1. OVERALL KPIs
# ============================================================

total_units = Profitability["units_sold"].sum()
total_revenue = Profitability["revenue"].sum()
total_cogs = Profitability["cogs"].sum()
total_waste_cost = Profitability["waste_cost"].sum()
total_gross_profit = Profitability["gross_profit"].sum()
total_adjusted_profit = Profitability["adjusted_profit"].sum()

gross_margin = (
    total_gross_profit / total_revenue * 100
    if total_revenue != 0 else 0
)

adjusted_margin = (
    total_adjusted_profit / total_revenue * 100
    if total_revenue != 0 else 0
)

print("\n" + "=" * 50)
print("OVERALL BUSINESS KPIs")
print("=" * 50)

print(f"Total Units Sold       : {total_units:,.0f}")
print(f"Total Revenue          : {total_revenue:,.2f}")
print(f"Total COGS             : {total_cogs:,.2f}")
print(f"Total Waste Cost       : {total_waste_cost:,.2f}")
print(f"Gross Profit           : {total_gross_profit:,.2f}")
print(f"Adjusted Profit        : {total_adjusted_profit:,.2f}")
print(f"Gross Margin           : {gross_margin:.2f}%")
print(f"Adjusted Margin        : {adjusted_margin:.2f}%")

# ============================================================
# 2. PRODUCT PERFORMANCE
# ============================================================

product_performance = Profitability[
    [
        "product_name",
        "units_sold",
        "revenue",
        "cogs",
        "gross_profit",
        "gross_margin_pct",
        "waste_cost",
        "adjusted_profit",
        "adjusted_margin_pct"
    ]
].copy()

print("\n" + "=" * 50)
print("TOP 10 PRODUCTS BY ADJUSTED PROFIT")
print("=" * 50)

print(
    product_performance
    .sort_values("adjusted_profit", ascending=False)
    .head(10)
    .to_string(index=False)
)

print("\n" + "=" * 50)
print("TOP 10 PRODUCTS BY REVENUE")
print("=" * 50)

print(
    product_performance
    .sort_values("revenue", ascending=False)
    .head(10)
    .to_string(index=False)
)

# ============================================================
# 3. MARGIN ANALYSIS
# ============================================================

print("\n" + "=" * 50)
print("HIGHEST ADJUSTED MARGIN PRODUCTS")
print("=" * 50)

print(
    product_performance
    .sort_values("adjusted_margin_pct", ascending=False)
    .head(10)
    .to_string(index=False)
)

print("\n" + "=" * 50)
print("LOWEST ADJUSTED MARGIN PRODUCTS")
print("=" * 50)

print(
    product_performance
    .sort_values("adjusted_margin_pct")
    .head(10)
    .to_string(index=False)
)

# ============================================================
# 4. WASTE IMPACT
# ============================================================

print("\n" + "=" * 50)
print("TOP 10 PRODUCTS BY WASTE COST")
print("=" * 50)

print(
    product_performance
    .sort_values("waste_cost", ascending=False)
    .head(10)
    .to_string(index=False)
)

# ============================================================
# 5. LOSS-MAKING PRODUCTS
# ============================================================

loss_making = product_performance[
    product_performance["adjusted_profit"] < 0
]

print("\n" + "=" * 50)
print("LOSS-MAKING PRODUCTS")
print("=" * 50)

if len(loss_making) > 0:
    print(loss_making.to_string(index=False))
else:
    print("No loss-making products identified.")

# ============================================================
# 6. DASHBOARD KPI TABLE
# ============================================================

kpis = pd.DataFrame({
    "metric": [
        "Total Units Sold",
        "Total Revenue",
        "Total COGS",
        "Total Waste Cost",
        "Gross Profit",
        "Adjusted Profit",
        "Gross Margin",
        "Adjusted Margin"
    ],
    "value": [
        total_units,
        total_revenue,
        total_cogs,
        total_waste_cost,
        total_gross_profit,
        total_adjusted_profit,
        gross_margin,
        adjusted_margin
    ]
})

print("\n" + "=" * 50)
print("DASHBOARD KPI TABLE")
print("=" * 50)

print(kpis.to_string(index=False))

print("\nStage 5 KPI analysis completed successfully.")
```

## WASTE IMPACT & MANAGEMENT INSIGHTS

```python
# ============================================================
# STAGE 5 - WASTE IMPACT & MANAGEMENT INSIGHTS
# ============================================================

# Waste as percentage of revenue
product_performance["waste_pct"] = (
    product_performance["waste_cost"]
    / product_performance["revenue"] * 100
)

# Profit lost because of waste
product_performance["profit_lost_to_waste"] = (
    product_performance["waste_cost"]
)

# ------------------------------------------------------------
# 1. HIGHEST WASTE RATE
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("PRODUCTS WITH HIGHEST WASTE RATE")
print("=" * 60)

print(
    product_performance[
        [
            "product_name",
            "revenue",
            "waste_cost",
            "waste_pct",
            "adjusted_profit",
            "adjusted_margin_pct"
        ]
    ]
    .sort_values("waste_pct", ascending=False)
    .head(10)
    .to_string(index=False)
)

# ------------------------------------------------------------
# 2. HIGHEST PROFIT LOSS FROM WASTE
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("PRODUCTS WITH HIGHEST PROFIT LOSS FROM WASTE")
print("=" * 60)

print(
    product_performance[
        [
            "product_name",
            "gross_profit",
            "waste_cost",
            "adjusted_profit"
        ]
    ]
    .sort_values("waste_cost", ascending=False)
    .head(10)
    .to_string(index=False)
)

# ------------------------------------------------------------
# 3. HIGH REVENUE + HIGH WASTE
# ------------------------------------------------------------

revenue_threshold = product_performance["revenue"].median()
waste_threshold = product_performance["waste_cost"].median()

high_revenue_high_waste = product_performance[
    (product_performance["revenue"] >= revenue_threshold) &
    (product_performance["waste_cost"] >= waste_threshold)
]

print("\n" + "=" * 60)
print("HIGH-REVENUE / HIGH-WASTE PRODUCTS")
print("=" * 60)

print(
    high_revenue_high_waste[
        [
            "product_name",
            "revenue",
            "waste_cost",
            "waste_pct",
            "adjusted_profit",
            "adjusted_margin_pct"
        ]
    ]
    .sort_values("waste_cost", ascending=False)
    .to_string(index=False)
)

# ------------------------------------------------------------
# 4. HIGH PROFIT + LOW WASTE
# ------------------------------------------------------------

profit_threshold = product_performance["adjusted_profit"].median()
low_waste_threshold = product_performance["waste_cost"].median()

high_profit_low_waste = product_performance[
    (product_performance["adjusted_profit"] >= profit_threshold) &
    (product_performance["waste_cost"] <= low_waste_threshold)
]

print("\n" + "=" * 60)
print("HIGH-PROFIT / LOW-WASTE PRODUCTS")
print("=" * 60)

print(
    high_profit_low_waste[
        [
            "product_name",
            "revenue",
            "waste_cost",
            "adjusted_profit",
            "adjusted_margin_pct"
        ]
    ]
    .sort_values("adjusted_profit", ascending=False)
    .to_string(index=False)
)

# ------------------------------------------------------------
# 5. MANAGEMENT RECOMMENDATIONS
# ------------------------------------------------------------

recommendations = pd.DataFrame({
    "Priority": [
        "Critical",
        "High",
        "High",
        "Growth",
        "Growth"
    ],
    "Product": [
        "Cream Cake Slice",
        "Chicken Pie",
        "Coconut Bread",
        "Puff Puff Pack",
        "Chin Chin Pack"
    ],
    "Action": [
        "Review pricing, production cost and waste immediately",
        "Reduce production waste and improve demand forecasting",
        "Reduce waste and review production quantities",
        "Increase focus on this high-margin product",
        "Consider increasing production/sales focus"
    ]
})

print("\n" + "=" * 60)
print("MANAGEMENT RECOMMENDATIONS")
print("=" * 60)

print(recommendations.to_string(index=False))

print("\nStage 5 business insight analysis completed.")
```
